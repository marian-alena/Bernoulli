```python
%%timeit -n10 -r10
from fractions import Fraction


def generate_bernoulli(n: int) -> list[Fraction]:
    """
    Akiyama–Tanigawa Algorithm
    Generates Bernoulli numbers B_0 through B_n as exact fractions.
    """
    a = [Fraction(1, m + 1) for m in range(n + 1)]
    bernoulli = []

    for m in range(n + 1):
        bernoulli.append(a[0])
        for j in range(n - m):
            a[j] = (j + 1) * (a[j] - a[j + 1])

    return bernoulli


n = 100
b = generate_bernoulli(n)
for index, value in enumerate(b):
    print(f"B_{index} = {value}")
```

```python
%%timeit -n10 -r10
from fractions import Fraction
import math


def bernoulli_recursive(n: int) -> list[Fraction]:
    """Generates B_0 through B_n (with B_1 = -1/2) using memoized recursion."""
    b = [Fraction(0)] * (n + 1)
    b[0] = Fraction(1)

    for m in range(1, n + 1):
        total = sum(math.comb(m + 1, k) * b[k] for k in range(m))
        b[m] = -Fraction(total, m + 1)

    return b


n = 100
b = bernoulli_recursive(n)
for index, value in enumerate(b):
    print(f"B_{index} = {value}")
```

```python
%%timeit -n10 -r10
import sympy

b = []
n = 100
for i in range(n + 1):
    b.append( sympy.bernoulli(i))
for index, value in enumerate(b):
    print(f"B_{index} = {value}")
```
